# Send ctDNA ORU R01 Reports

Sends the 100 example ctDNA report messages generated in `Input/V2/R01/ctdna{NHSNumber}_{seq}.txt` to the v2 server.

```python
import glob
import os
import requests
from dotenv import load_dotenv

load_dotenv()

v2Server = os.getenv("V2_SERVER")
```

```python
report_list = sorted(
    os.path.basename(f)
    for f in glob.glob('Input/V2/R01/ctdna*_[0-9][0-9][0-9].txt')
)
len(report_list)
```

```python
for file in report_list:
    with open('Input/V2/R01/' + file, 'rb') as g:
        s = g.read()

        r = requests.post(v2Server, data = s)
        print(file)
        print(r.status_code)
        print(r.text)
```
